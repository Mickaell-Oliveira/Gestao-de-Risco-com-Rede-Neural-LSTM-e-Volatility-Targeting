Reprodução da melhor estratégia do estudo anterior (Meta-Estratégia Heston-GARCH) aplicada ao S&P 500 e ao Ibovespa, consumindo diretamente os módulos em `src/`.


```python
import os
import sys
import pandas as pd

# Acesso aos módulos locais
sys.path.append(os.path.abspath('..'))

from src import (
    ASSET_PARAMS,
    MarketDataLoader,
    RiskIndicators,
    Backtester,
    ResultPlotter
)

loader = MarketDataLoader()
indicators = RiskIndicators()
backtester = Backtester()
plotter = ResultPlotter()
```

## 1. Pipeline S&P 500 (`^GSPC`)


```python
# Carga e indicadores
sp_raw = loader.load('^GSPC')
sp_df = indicators.compute(sp_raw, ASSET_PARAMS['^GSPC'])

# Simulação e resumo
sp_eval, sp_res = backtester.run(sp_df)
sp_table = backtester.summarize(sp_res, 'S&P 500')
sp_table
```

## 2. Pipeline Ibovespa (`^BVSP`)


```python
# Carga e indicadores
ib_raw = loader.load('^BVSP')
ib_df = indicators.compute(ib_raw, ASSET_PARAMS['^BVSP'])

# Simulação e resumo
ib_eval, ib_res = backtester.run(ib_df)
ib_table = backtester.summarize(ib_res, 'Ibovespa')
ib_table
```

## 3. Tabela Comparativa (2018 - 2025)


```python
# Consolidação das métricas
comparison = pd.concat([sp_table, ib_table], ignore_index=True)
comparison
```

## 4. Gráficos


```python
# Exibição e exportação
plotter.plot_comparison(sp_res, ib_res, show=True)
plotter.plot_panic_override(ib_eval, ib_res, show=True)
```
